```python
from langchain_core.tools import tool, StructuredTool, BaseTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a: float = Field(..., description="The first number to multiply")
    b: float = Field(..., description="The second number to multiply")
```

```python
def multiply_func(a : float, b : float):
    return a * b
```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    name="multiply",
    arguments_schema=MultiplyInput,
    description="Multiplies two numbers together."
)
```

```python
res = multiply_tool.invoke({"a": 5, "b": 10})
print(res)  # Output: 50
```

<h1>BaseTool</h1>

```python
class MultiplyTool(BaseTool):
    name : str = "multiply"
    description : str = "Multiplies two numbers together."
    arguments_schema : type = MultiplyInput

    def _run(self, a: float, b: float) -> float:
        return a * b
```

```python
multiply_tool = MultiplyTool()

res = multiply_tool.invoke({"a": 5, "b": 10})
print(res)  # Output: 50
```
